# Large RPA Motifs — SN5.2 Small Networks with Novel Topological Features

Six small networks from Figure 8 of Araujo & Liotta (2018), *Nature Communications*, each incorporating novel RPA topological features not present in minimal 3-node circuits. For each case the steady-state of the output node **O** is derived symbolically with SymPy and parameters are partitioned into those that affect O* and those that do not.

```python
from sympy import symbols, solve, simplify, diff, latex
from IPython.display import display, Math

def analyze(label, eqs, state_vars, output_sym, params):
    sol = solve(eqs, state_vars, dict=True)
    if not sol:
        print(f"Case {label}: no closed-form solution found")
        return None
    O_ss = simplify(sol[0][output_sym])
    print(f"Case {label} — O* =")
    display(Math(r'O^* = ' + latex(O_ss)))
    names = [str(p) for p in params]
    no_eff = [n for n, p in zip(names, params) if diff(O_ss, p) == 0]
    affects = [n for n, p in zip(names, params) if diff(O_ss, p) != 0]
    print(f"  No effect on O*: {', '.join(no_eff)}")
    print(f"  Affects O*:      {', '.join(affects)}")
    return O_ss
```

## Case a — Opposer module (upstream) + Balancer module (downstream)

$$\dot{I} = k_1 \cdot \text{Input} - k_2 I - k_3 O_1$$
$$\dot{O} = k_4 I + k_5 C - k_6 O$$
$$\dot{O}_1 = k_7 I - k_8$$
$$\dot{C} = k_9 O_1 - k_{10} B C$$
$$\dot{B} = k_{11} O_1 - k_{12} B$$

Paper parameters: $k_1 = k_3 = k_6 = k_7 = 1$; $k_2 = k_4 = k_5 = k_8 = k_9 = k_{10} = k_{11} = k_{12} = 0.5$

```python
Input = symbols('Input', positive=True)
Iv, O, O1, C, B = symbols('I O O1 C B', positive=True)
k1,k2,k3,k4,k5,k6,k7,k8,k9,k10,k11,k12 = symbols(
    'k1 k2 k3 k4 k5 k6 k7 k8 k9 k10 k11 k12', positive=True)

eqs_a = [k1*Input - k2*Iv - k3*O1,
         k4*Iv + k5*C - k6*O,
         k7*Iv - k8,
         k9*O1 - k10*B*C,
         k11*O1 - k12*B]

O_a = analyze('a', eqs_a, [Iv, O, O1, C, B], O,
              [Input, k1,k2,k3,k4,k5,k6,k7,k8,k9,k10,k11,k12])
```

## Case b — Balancer module (upstream) + Opposer module (downstream)

$$\dot{I} = k_1 \cdot \text{Input} - k_2 I$$
$$\dot{B} = k_3 I - k_4 B$$
$$\dot{X} = \frac{k_5 B}{O_1} - k_7 X$$
$$\dot{O}_1 = k_8 X - k_9$$
$$\dot{O} = k_{10} I X - k_{11} O B$$

Paper parameters: $k_1 = k_2 = k_3 = k_4 = k_7 = k_8 = k_{10} = k_{11} = 1$; $k_5 = k_9 = 0.5$

```python
Input = symbols('Input', positive=True)
Iv, B, X, O1, O = symbols('I B X O1 O', positive=True)
k1,k2,k3,k4,k5,k7,k8,k9,k10,k11 = symbols(
    'k1 k2 k3 k4 k5 k7 k8 k9 k10 k11', positive=True)

eqs_b = [k1*Input - k2*Iv,
         k3*Iv - k4*B,
         k5*B/O1 - k7*X,
         k8*X - k9,
         k10*Iv*X - k11*O*B]

O_b = analyze('b', eqs_b, [Iv, B, X, O1, O], O,
              [Input, k1,k2,k3,k4,k5,k7,k8,k9,k10,k11])
```

## Case c — Two Opposer modules in series

$$\dot{I} = k_1 \cdot \text{Input} - k_2 I - k_3 O_1$$
$$\dot{O} = k_4 I + k_5 X - k_6 O$$
$$\dot{O}_1 = k_7 I - k_8$$
$$\dot{X} = k_9 O_1 - k_{10} X - k_{11} O_2$$
$$\dot{O}_2 = k_{12} X - k_{13}$$

Paper parameters: $k_1 = k_3 = k_6 = k_7 = k_{12} = 1$; $k_2 = k_4 = k_5 = k_8 = k_9 = k_{10} = k_{11} = k_{13} = 0.5$

```python
Input = symbols('Input', positive=True)
Iv, O, O1, X, O2 = symbols('I O O1 X O2', positive=True)
k1,k2,k3,k4,k5,k6,k7,k8,k9,k10,k11,k12,k13 = symbols(
    'k1 k2 k3 k4 k5 k6 k7 k8 k9 k10 k11 k12 k13', positive=True)

eqs_c = [k1*Input - k2*Iv - k3*O1,
         k4*Iv + k5*X - k6*O,
         k7*Iv - k8,
         k9*O1 - k10*X - k11*O2,
         k12*X - k13]

O_c = analyze('c', eqs_c, [Iv, O, O1, X, O2], O,
              [Input, k1,k2,k3,k4,k5,k6,k7,k8,k9,k10,k11,k12,k13])
```

## Case d — Single Opposer module with two-node opposing set

$$\dot{I} = k_1 \cdot \text{Input} - k_2 I + k_3 O_1$$
$$\dot{O} = k_4 I - k_5 O$$
$$\dot{O}_1 = k_{11} X - k_{12}$$
$$\dot{X} = k_8 O_2 - k_9 X - k_{10} O_1$$
$$\dot{O}_2 = k_6 O - k_7 X$$

Paper parameters: $k_1 = k_2 = k_3 = k_9 = k_{12} = 0.5$; $k_4 = k_5 = k_6 = k_7 = k_8 = k_{10} = k_{11} = 1$

```python
Input = symbols('Input', positive=True)
Iv, O, O1, X, O2 = symbols('I O O1 X O2', positive=True)
k1,k2,k3,k4,k5,k6,k7,k8,k9,k10,k11,k12 = symbols(
    'k1 k2 k3 k4 k5 k6 k7 k8 k9 k10 k11 k12', positive=True)

eqs_d = [k1*Input - k2*Iv + k3*O1,
         k4*Iv - k5*O,
         k11*X - k12,
         k8*O2 - k9*X - k10*O1,
         k6*O - k7*X]

O_d = analyze('d', eqs_d, [Iv, O, O1, X, O2], O,
              [Input, k1,k2,k3,k4,k5,k6,k7,k8,k9,k10,k11,k12])
```

## Case e — Opposer module (upstream, two-node set) + Balancer module (downstream)

$$\dot{I} = k_1 \cdot \text{Input} - k_2 O_1 I$$
$$\dot{O}_1 = k_4 I - k_5 X$$
$$\dot{X} = k_6 O_1 - k_7 O_2 - k_8 X$$
$$\dot{O}_2 = k_9 X - k_{10}$$
$$\dot{B} = k_{11} O_2 - k_{12} B$$
$$\dot{C} = k_{14} O_2 - k_{15} B C$$
$$\dot{O} = k_{16} I + k_{17} C - k_{18} O$$

Paper parameters: $k_2{=}k_4{=}k_5{=}k_6{=}k_7{=}k_9{=}k_{11}{=}k_{12}{=}k_{15}{=}k_{18}{=}1$; $k_1{=}k_{10}{=}k_{14}{=}k_{16}{=}k_{17}{=}0.5$; $k_8{=}0.4$

```python
Input = symbols('Input', positive=True)
Iv, O1, X, O2, B, C, O = symbols('I O1 X O2 B C O', positive=True)
k1,k2,k4,k5,k6,k7,k8,k9,k10,k11,k12,k14,k15,k16,k17,k18 = symbols(
    'k1 k2 k4 k5 k6 k7 k8 k9 k10 k11 k12 k14 k15 k16 k17 k18', positive=True)

eqs_e = [k1*Input - k2*O1*Iv,
         k4*Iv - k5*X,
         k6*O1 - k7*O2 - k8*X,
         k9*X - k10,
         k11*O2 - k12*B,
         k14*O2 - k15*B*C,
         k16*Iv + k17*C - k18*O]

O_e = analyze('e', eqs_e, [Iv, O1, X, O2, B, C, O], O,
              [Input, k1,k2,k4,k5,k6,k7,k8,k9,k10,k11,k12,k14,k15,k16,k17,k18])
```

## Case f — Opposer module (upstream, two-node set) + Opposer module (downstream, single node)

$$\dot{I} = k_1 \cdot \text{Input} - k_2 I - k_3 O_1$$
$$\dot{O}_1 = k_4 I - k_5 X_1$$
$$\dot{X}_1 = k_6 O_1 - k_7 O_2 - k_8 X_1$$
$$\dot{O}_2 = k_9 X_1 - k_{10}$$
$$\dot{X}_2 = k_{11} O_2 - k_{12} X_2 - k_{13} O_3$$
$$\dot{O}_3 = k_{14} X_2 - k_{15}$$
$$\dot{O} = k_{16} I + k_{17} X_2 - k_{18} O$$

Paper parameters: $k_1{=}k_2{=}k_3{=}k_{10}{=}k_{12}{=}k_{15}{=}k_{16}{=}k_{17}{=}0.5$; $k_4{=}k_5{=}k_6{=}k_7{=}k_8{=}k_9{=}k_{11}{=}k_{13}{=}k_{14}{=}k_{18}{=}1$

```python
Input = symbols('Input', positive=True)
Iv, O1, X1, O2, X2, O3, O = symbols('I O1 X1 O2 X2 O3 O', positive=True)
k1,k2,k3,k4,k5,k6,k7,k8,k9,k10,k11,k12,k13,k14,k15,k16,k17,k18 = symbols(
    'k1 k2 k3 k4 k5 k6 k7 k8 k9 k10 k11 k12 k13 k14 k15 k16 k17 k18', positive=True)

eqs_f = [k1*Input - k2*Iv - k3*O1,
         k4*Iv - k5*X1,
         k6*O1 - k7*O2 - k8*X1,
         k9*X1 - k10,
         k11*O2 - k12*X2 - k13*O3,
         k14*X2 - k15,
         k16*Iv + k17*X2 - k18*O]

O_f = analyze('f', eqs_f, [Iv, O1, X1, O2, X2, O3, O], O,
              [Input, k1,k2,k3,k4,k5,k6,k7,k8,k9,k10,k11,k12,k13,k14,k15,k16,k17,k18])
```
