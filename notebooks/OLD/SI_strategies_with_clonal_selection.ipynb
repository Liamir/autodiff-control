```python
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.integrate import solve_ivp
```

```python
def clonal_selection_rhs(t, y, r_s, r_r, k_kill, carrying_capacity, therapy_on):
    """two-population logistic competition; therapy only kills sensitive cells."""
    n_s, n_r = y
    assert carrying_capacity > 0
    assert therapy_on in (0, 1)

    total = n_s + n_r
    growth_factor = 1 - total / carrying_capacity

    dn_s = r_s * n_s * growth_factor - therapy_on * k_kill * n_s
    dn_r = r_r * n_r * growth_factor
    return [dn_s, dn_r]
```

```python
def simulate_clonal_selection(
    n_s0, n_r0, r_s, r_r, k_kill, carrying_capacity, therapy_on, t_max, n_points=200
):
    assert n_s0 >= 0 and n_r0 >= 0, "populations must start non-negative"
    assert r_s > 0 and r_r > 0, "growth rates must be positive"

    t_eval = np.linspace(0, t_max, n_points)
    sol = solve_ivp(
        clonal_selection_rhs,
        t_span=(0, t_max),
        y0=[n_s0, n_r0],
        args=(r_s, r_r, k_kill, carrying_capacity, therapy_on),
        t_eval=t_eval,
        method="RK45",
    )
    assert sol.success, f"integration failed: {sol.message}"
    return sol.t, sol.y[0], sol.y[1]
```

```python
def plot_clonal_selection(t, n_s, n_r, title=""):
    fig, ax = plt.subplots(figsize=(6, 3))
    ax.plot(t, n_s, label="cs (sensitive)")
    ax.plot(t, n_r, label="cr (resistant)")
    ax.set_xlabel("time")
    ax.set_ylabel("population")
    ax.set_title(title)
    sns.despine(ax=ax)
    ax.legend(frameon=False, bbox_to_anchor=(1, 1))
    return fig, ax

def plot_population_proportions(t, n_s, n_r, title=""):
    total = n_s + n_r
    assert np.all(total > 0), "total population must be positive to compute proportions"

    prop_s = n_s / total
    prop_r = n_r / total

    fig, ax = plt.subplots(figsize=(6, 3))
    ax.plot(t, prop_s, label="cs (sensitive)")
    ax.plot(t, prop_r, label="cr (resistant)")
    ax.set_xlabel("time")
    ax.set_ylabel("proportion")
    ax.set_ylim(0, 1)
    ax.set_title(title)
    sns.despine(ax=ax)
    ax.legend(frameon=False, bbox_to_anchor=(1, 1))
    return fig, ax
```

```python
# under therapy, sensitive cells should decline and resistant should take over
t, n_s, n_r = simulate_clonal_selection(
    n_s0=10, n_r0=1, r_s=0.5, r_r=0.2, k_kill=0.6,
    carrying_capacity=1000, therapy_on=1, t_max=100,
)
assert n_r[-1] > n_s[-1], "resistant cells should dominate under sustained therapy"
plot_clonal_selection(t, n_s, n_r, title="therapy on")
plot_population_proportions(t, n_s, n_r, title="proportion under therapy")
```

## Combining the ABC timescale-separation strategy with two competing clones

The ABC model (`figure4 copy.ipynb`) exploits a homeostatic $A$–$B$ pair that always relaxes back to its off-treatment fixed point $B^*=\mu$, regardless of whether the drug is on or off (a robust-perfect-adaptation property). Net tumor decay only accumulates while $B$ is transiently pushed below $\mu$, so letting $A,B$ reset during $T_{off}$ before pulsing again extracts more cumulative kill than one continuous dose of the same total duration.

Question: **does this still work once the tumor is not a single population but two competing clones?** We replace the single tumor variable $C$ with the clonal-selection pair from above — a drug-**sensitive** clone $C_s$ killed while $B<\mu$ (same mechanism as $C$), and a drug-**insensitive resistant** clone $C_r$ that only competes with $C_s$ for one shared carrying capacity:

$$\dot A = a_1(\mu - B), \qquad \dot B = b_1(t)\,A - b_2 B$$
$$\dot C_s = r_s C_s\Bigl(1-\tfrac{C_s+C_r}{K}\Bigr) - c_2 C_s(\mu-B)^+, \qquad \dot C_r = r_r C_r\Bigl(1-\tfrac{C_s+C_r}{K}\Bigr)$$

Crucially, $A,B$ have **no dependence on $C_s,C_r$** — so whatever pulse schedule the timescale-separation argument picks, it produces exactly the same kill dynamics on $C_s$ whether or not a second clone exists. The only new ingredient is $C_r$: it competes with $C_s$ for capacity, but the treatment never touches it directly. The model is `ABCClonalModel` in `rpa_control/models.py`, compatible with the same `simulate_rpa_on_off` / `simulate_rpa_intermittent` helpers used in `sot_nfl_iffl_demo.ipynb`.

```python
from rpa_control.models import ABCClonalModel, b1, b1_on, A_ss, B_ss
from rpa_control import simulation

print(f'A_ss={A_ss:.3f}  B_ss={B_ss:.3f}  b1 (off)={b1}  b1_on (on)={b1_on}')
```

### Deriving the schedule

We reuse the exact growth-vs-decay argument from `figure4 copy.ipynb` Panel C: the decay integral $S(T_{on})$ depends only on the AB parameters (unchanged from before), and the per-capita growth rate of $C_s$ while the population is far below carrying capacity is just $r_s$. Setting $r_s = c_1 = 0.05$ — the same slow growth rate figure4 used — and $T_{off}=10$, we derive $T_{on}^{opt}$ once, and reuse it for every scenario below.

```python
a1_, b2_, mu_, c2_ = 1.0, 2.0, 1.0, 1.0   # AB/kill parameters, same as the ABC model
disc = b2_**2 - 4 * a1_ * b1_on
lam1 = (-b2_ + np.sqrt(disc)) / 2
lam2 = (-b2_ - np.sqrt(disc)) / 2
K1   = b2_ * mu_ * (b1_on / b1 - 1) / (lam1 - lam2)

def decay_analytic(Ton):
    """Analytic cumulative log-decay of Cs over one on-phase (independent of Cr)."""
    Ton = np.asarray(Ton)
    return -c2_ * K1 * (
        np.exp(lam1 * Ton) / lam1 - np.exp(lam2 * Ton) / lam2 - 1 / lam1 + 1 / lam2
    )

r_s_slow = 0.05   # same magnitude as figure4's c1: growth timescale >> AB recovery timescale
T_off    = 10.0

Ton_arr = np.linspace(0.01, 40, 2000)
net_arr = decay_analytic(Ton_arr) - r_s_slow * (Ton_arr + T_off)
Ton_opt = Ton_arr[np.argmax(net_arr)]
print(f'Ton_opt = {Ton_opt:.3f}  (decay={decay_analytic(Ton_opt):.3f}, '
     f'growth={r_s_slow*(Ton_opt+T_off):.3f}, net={net_arr.max():.3f} per cycle)')
```

```python
def plot_abc_clonal_run(t, y, on_ranges, title='', thresholds=None):
    """One figure: A, B, {Cs, Cr, total} on shared log axes, and resistant fraction.

    on_ranges: list of (t_start, t_end) treatment-on intervals to shade
    (fixed-period schedules and threshold-driven schedules both produce this).
    thresholds: optional list of (label, value) pairs -- drawn as horizontal
    reference lines on the population panel (e.g. PSA-driven's on/off burden
    triggers).
    """
    A, B, Cs, Cr = y
    total = Cs + Cr
    resist_frac = Cr / total

    fig, axes = plt.subplots(4, 1, figsize=(8, 8), sharex=True,
                             gridspec_kw={'hspace': 0.1, 'height_ratios': [1, 1, 1.6, 1]})
    for ax in axes:
        for t0, t1 in on_ranges:
            ax.axvspan(t0, t1, color='lightgray', alpha=0.4, lw=0)
        sns.despine(ax=ax)

    axes[0].plot(t, A, color='C1', lw=1.4)
    axes[0].set_ylabel('$A$', fontsize=9)

    axes[1].plot(t, B, color='C0', lw=1.4)
    axes[1].axhline(mu_, color='C0', ls='--', lw=0.9, alpha=0.5, label=r'$\mu$')
    axes[1].set_ylabel('$B$', fontsize=9)
    axes[1].legend(fontsize=7.5, frameon=False, loc='lower right')

    axes[2].plot(t, Cs, label='$C_s$ (sensitive)')
    axes[2].plot(t, Cr, label='$C_r$ (resistant)')
    axes[2].plot(t, total, color='k', ls='--', lw=1.2, label='total')
    if thresholds:
        for label, val in thresholds:
            axes[2].axhline(val, color='0.4', ls=':', lw=1.2, alpha=0.85, label=label)
    axes[2].set_ylabel('population')
    axes[2].set_yscale('log')
    axes[2].legend(fontsize=7.5, frameon=False, loc='center right')

    axes[3].plot(t, resist_frac, color='C3')
    axes[3].set_ylabel('resistant\nfraction')
    axes[3].set_ylim(0, 1)
    axes[3].set_xlabel('time')

    axes[0].set_title(title, fontsize=10)
    plt.tight_layout()
    plt.show()


def periodic_on_ranges(Ton, Toff, n_cycles):
    return [(c * (Ton + Toff), c * (Ton + Toff) + Ton) for c in range(n_cycles)]


def plot_quantity_comparison(runs, extract, ylabel, title='', yscale=None, thresholds=None, ylim=None):
    """runs: list of (label, t, y) tuples, e.g. different strategies on the
    same scenario. extract(y) -> 1D array to plot. One line per run, all on
    a single pair of axes, for direct side-by-side comparison.
    thresholds: optional list of (label, value) pairs -- drawn as horizontal
    reference lines (e.g. PSA-driven's on/off burden triggers).
    ylim: optional (lo, hi) tuple, e.g. (0, 1) for a fraction."""
    fig, ax = plt.subplots(figsize=(8, 3.5))
    for label, t, y in runs:
        ax.plot(t, extract(y), label=label, lw=1.6)
    if thresholds:
        for label, val in thresholds:
            ax.axhline(val, color='0.4', ls=':', lw=1.2, alpha=0.85, label=label)
    if yscale:
        ax.set_yscale(yscale)
    if ylim:
        ax.set_ylim(*ylim)
    ax.set_ylabel(ylabel)
    ax.set_xlabel('time')
    ax.legend(fontsize=8, frameon=False)
    sns.despine(ax=ax)
    ax.set_title(title, fontsize=10)
    plt.tight_layout()
    plt.show()
```

```python
from scipy.optimize import brentq

def time_to_progression(t, total, baseline, multiple=2.0):
    """First time `total` crosses `multiple` x `baseline`, or None if it never does."""
    threshold = multiple * baseline
    above = total >= threshold
    if not np.any(above):
        return None
    idx = np.argmax(above)   # first index where total >= threshold
    if idx == 0:
        return t[0]           # already at/above threshold at t=0
    f = lambda tt: np.interp(tt, t, total) - threshold
    return brentq(f, t[idx - 1], t[idx])

burden_fn = lambda y: y[2] + y[3]   # PSA proxy: total tumor size Cs + Cr
```

### An edge case: where PSA-driven should fail and TST should succeed

Before the resistance-level scenarios below, consider a case engineered to make the mismatch between the two dosing philosophies as stark as possible: **slow cancer shrinkage** (small $c_2$) and **slow cancer growth** (small $r_s, r_r$) — i.e. tumor dynamics that are much slower than the fast $A,B$ dynamics underneath them.

Scale $c_2$, $r_s$, and $r_r$ all down by the same factor $k=0.1$ from the values used throughout the rest of this notebook ($c_2:1.0\to0.1$; $r_s:0.05\to0.005$; $r_r:0.03\to0.003$). Scaling $c_2$ and $r_s$ by the *same* factor leaves `Ton_opt` exactly unchanged — the objective being maximized, $c_2\cdot g(T_{on}) - r_s\cdot(T_{on}+T_{off})$, just gets multiplied by $k$ overall, which doesn't move its argmax. So **TST's schedule doesn't need to be re-derived at all** — it's set entirely by the fast AB timescale, never by the tumor's own.

Why this should specifically break PSA: the maximum log-decay obtainable from *any single, arbitrarily long* dosing episode is bounded by

$$S_\infty = c_2\cdot\frac{b_2\mu(1-b_{1,on}/b_{1,off})}{\lambda_1\lambda_2}$$

At $c_2=0.1$, $S_\infty=0.2$. PSA-driven only switches off once accumulated decay reaches $\ln(1/\text{off\_frac})$ — for the default `off_frac=0.5` that's $\ln 2\approx0.693 \gg S_\infty$. **That's impossible to reach via one continuous dose, however long it runs** — so PSA-driven should get stuck permanently ON, degenerating into exactly the naive continuous-dosing strategy that pulsing is supposed to beat. TST, by re-triggering a fresh transient every cycle regardless of how slowly the tumor itself evolves, keeps accumulating decay indefinitely instead.

```python
# Same base rates used throughout this notebook (r_s=0.05, r_r=0.03, c2=1.0),
# scaled down by the same factor k -- this leaves Ton_opt unchanged (see above).
k_edge   = 0.1
c2_edge  = k_edge * 1.0
r_s_edge = k_edge * 0.05
r_r_edge = k_edge * 0.03

model_edge = ABCClonalModel(r_s=r_s_edge, r_r=r_r_edge, carrying_capacity=1000.0,
                            cr_sensitivity=0.0, c2=c2_edge)

S_inf_edge    = -c2_edge * K1 * (-1 / lam1 + 1 / lam2)
off_frac_edge = 0.5
print(f'r_s={r_s_edge}, r_r={r_r_edge}, c2={c2_edge}')
print(f'S_inf = {S_inf_edge:.3f}  vs.  ln(1/off_frac) = {np.log(1/off_frac_edge):.3f}  '
     f'--> PSA\'s off-trigger is '
     f'{"reachable" if S_inf_edge > np.log(1/off_frac_edge) else "UNREACHABLE via continuous dosing"}')
```

```python
Cs0_edge, Cr0_edge = 10.0, 0.01
y0_edge          = np.array([A_ss, B_ss, Cs0_edge, Cr0_edge])
baseline_edge    = Cs0_edge + Cr0_edge
T_horizon_edge   = 300.0
n_cycles_edge    = int(T_horizon_edge // (Ton_opt + T_off))
on_ranges_edge   = periodic_on_ranges(Ton_opt, T_off, n_cycles_edge)

# TST: the exact same periodic schedule used everywhere else in this notebook
t_tst_edge, y_tst_edge = simulation.simulate_rpa_intermittent(
    model_edge, 'b1', b1_on, b1, t_on=Ton_opt, t_off=T_off,
    n_cycles=n_cycles_edge, y0=y0_edge, plot=False)
plot_abc_clonal_run(t_tst_edge, y_tst_edge, on_ranges_edge,
                    title='Edge case, TST (periodic schedule)')

Cs_tst_edge, Cr_tst_edge = y_tst_edge[2], y_tst_edge[3]
total_tst_edge = Cs_tst_edge + Cr_tst_edge
ttp_tst_edge   = time_to_progression(t_tst_edge, total_tst_edge, baseline_edge)
print(f'TST:  min total = {total_tst_edge.min():.4f} at t={t_tst_edge[total_tst_edge.argmin()]:.1f}   '
     f'final total = {total_tst_edge[-1]:.4f}   time to progression = {ttp_tst_edge}')
```

```python
# PSA-driven: same off_frac=0.5, on_frac=1.0 convention used elsewhere
t_psa_edge, y_psa_edge, on_ranges_psa_edge = simulation.simulate_threshold_driven(
    model_edge, 'b1', b1_on, b1, burden_fn, off_frac=off_frac_edge, on_frac=1.0,
    T_horizon=T_horizon_edge, y0=y0_edge)
plot_abc_clonal_run(t_psa_edge, y_psa_edge, on_ranges_psa_edge,
                    title='Edge case, PSA-driven',
                    thresholds=[('off threshold', off_frac_edge * baseline_edge),
                                ('on threshold', 1.0 * baseline_edge)])

Cs_psa_edge, Cr_psa_edge = y_psa_edge[2], y_psa_edge[3]
total_psa_edge = Cs_psa_edge + Cr_psa_edge
ttp_psa_edge   = time_to_progression(t_psa_edge, total_psa_edge, baseline_edge)
duty_psa_edge  = sum(min(e, T_horizon_edge) - s for s, e in on_ranges_psa_edge) / T_horizon_edge
print(f'PSA-driven:  {len(on_ranges_psa_edge)} on-interval(s), duty = {duty_psa_edge:.2f}')
print(f'  min total = {total_psa_edge.min():.4f} at t={t_psa_edge[total_psa_edge.argmin()]:.1f}   '
     f'final total = {total_psa_edge[-1]:.4f}   time to progression = {ttp_psa_edge}')
```

#### Direct comparison: TST vs. PSA-driven (edge case)

```python
runs_edge = [('TST', t_tst_edge, y_tst_edge), ('PSA-driven', t_psa_edge, y_psa_edge)]
plot_quantity_comparison(runs_edge, lambda y: y[2] + y[3], 'total population',
                         title='Edge case: total population, TST vs. PSA-driven', yscale='log',
                         thresholds=[('off threshold', off_frac_edge * baseline_edge),
                                     ('on threshold', 1.0 * baseline_edge)])
plot_quantity_comparison(runs_edge, lambda y: y[1], '$B$',
                         title='Edge case: $B(t)$, TST vs. PSA-driven')
```

**Confirmed.** PSA-driven never once triggers its off-switch — one continuous on-interval for the entire horizon, exactly the degenerate "continuous dosing" behavior predicted. It gets a single bounded dip (to ~8.7, ~13% off baseline, at t≈9) and then relapses fast, crossing back above $2\times$ baseline by t≈181 and continuing to grow (to ≈36 by t=300). TST, applying the identical schedule it would use regardless of how slow the tumor is, keeps re-triggering fresh transients every cycle: it's still steadily declining at the end of this window, down to ≈2.7 (73% below baseline) by t≈286, with no sign yet of relapsing. This holds for essentially the whole practical range of `off_frac` (checked 0.3–0.85 on a longer run, all get stuck at duty=1.0); only a very shallow target (`off_frac=0.9`, needing less decay than $S_\infty$) lets PSA-driven adapt at all, and even then TST still wins.

(On a much longer horizon TST eventually relapses too, once its own slow growth catches up — but the point here is the contrast within any clinically reasonable follow-up window: PSA-driven has already relapsed while TST is still working.)

The lesson: PSA-driven's blindness to the fast $A,B$ dynamics isn't just a minor inefficiency — when the tumor's own timescale is much slower than $A,B$'s, it can make PSA-driven degenerate into the *worst* strategy (unmodulated continuous dosing) without the controller ever noticing, while TST's schedule is unaffected by the tumor's timescale entirely and keeps working exactly as designed.

### The mechanism: per-cycle log-odds drift between the two clones

Rather than comparing runs at different horizons, ask directly what happens to the tumor's *composition* $\ln(C_r/C_s)$ under repeated pulsing — and let the resistant clone's **drug sensitivity**, not its growth rate, be the free parameter. Let `cr_sensitivity`$\in[0,1]$ be the fraction of $C_s$'s kill rate that still applies to $C_r$ (0 = fully resistant/immune, as originally implemented; values in $(0,1)$ = partial resistance — still killed, just less efficiently). The composition's rate of change is

$$\frac{d}{dt}\ln\frac{C_r}{C_s} = \underbrace{\text{growth\_factor}(t)\cdot(r_r-r_s)}_{\text{favors }C_s\text{ (}r_s>r_r\text{)}} + \underbrace{c_2(\mu-B(t))^+\cdot(1-\text{cr\_sensitivity})}_{\text{favors }C_r\text{, scaled down as cr\_sensitivity grows}}$$

Integrating one on/off cycle, while the tumor is still far below carrying capacity (growth_factor≈1):

$$\Delta\ln(C_r/C_s) \text{ per cycle} \approx (1-\text{cr\_sensitivity})\,S(T_{on}) - (r_s-r_r)(T_{on}+T_{off})$$

$S(T_{on})$ is the same decay integral as before; $(r_s-r_r)$ is $C_s$'s fitness advantage, held fixed here at the original values ($r_s=0.05$, $r_r=0.03$). **This is the right knob for "level of resistance"** — not the fitness gap, which we tried previously and found doesn't actually change the qualitative picture (see below).

```python
net_abs = decay_analytic(Ton_opt) - r_s_slow * (Ton_opt + T_off)   # net absolute decay achieved (Part 1)
print(f'S(Ton_opt) = {decay_analytic(Ton_opt):.3f},  net absolute decay = {net_abs:+.3f}\n')

r_r_slow = 0.03   # reverted to the original fitness gap; held fixed for both scenarios below
print(f'Fixing r_s={r_s_slow}, r_r={r_r_slow}. Per-cycle Delta ln(Cr/Cs) as cr_sensitivity varies:')
for cr_sens in [0.0, 0.2, 0.5, 0.79, 0.95]:
    dlogodds = (1 - cr_sens) * decay_analytic(Ton_opt) - (r_s_slow - r_r_slow) * (Ton_opt + T_off)
    print(f'  cr_sensitivity={cr_sens:.2f}:  Delta ln(Cr/Cs) = {dlogodds:+.3f}')

cr_sens_crit = 1 - (r_s_slow - r_r_slow) * (Ton_opt + T_off) / decay_analytic(Ton_opt)
print(f'\ncritical cr_sensitivity where the sign flips: {cr_sens_crit:.3f}  '
     f'(above this, Cs\'s fitness advantage outweighs the reduced kill, and resistant '
     f'fraction would actually *decrease* every cycle -- not explored further here)')
```

At `cr_sensitivity=0` this reduces exactly to the old formula, and the same structural fact holds: whenever the schedule achieves net absolute decay (`net_abs>0`), $\Delta\ln(C_r/C_s)$ is guaranteed positive — full resistance is a foregone conclusion given long enough. That guarantee doesn't survive increasing `cr_sensitivity` indefinitely, though: past the critical value printed above (~0.79 here), the sign flips and resistant fraction would actually *decrease* every cycle, since a resistant clone that's still substantially killed can lose ground to the fitness-advantaged $C_s$ even while both are being suppressed.

We compare two levels of resistance directly, holding $r_s, r_r$, the schedule, and the initial condition fixed throughout (a rare pre-existing resistant seed, $C_r(0)=0.01$, 1/1000th of $C_s(0)$): **Scenario A**, $C_r$ completely unaffected by therapy (`cr_sensitivity=0`, the original assumption); and **Scenario B**, $C_r$ somewhat affected — still killed, just far less efficiently than $C_s$ (`cr_sensitivity=0.2`, well below the critical value) — while continuing to grow, just slowly. As before, we track **absolute burden** and **time to progression**, not just the final fraction, since fraction alone can't distinguish "a negligible residual population that happens to be 100% resistant" from "a fully relapsed, resistant-dominated tumor."

### Scenario A — fully resistant clone

$C_r$ is completely immune to the drug (`cr_sensitivity=0`, the original model behavior). This is also the reference scenario Part 2 below builds on — it's the one where clonal selection actually causes a relapse, making a scheduling-philosophy comparison meaningful.

```python
Cs0, Cr0 = 10.0, 0.01   # rare pre-existing resistant seed: 1/1000th of Cs0
y0_scenario = np.array([A_ss, B_ss, Cs0, Cr0])
baseline = Cs0 + Cr0

T_horizon_long = 600.0   # shared horizon for both scenarios (name kept for Part 2 below)
n_cycles_long  = int(T_horizon_long // (Ton_opt + T_off))
on_ranges_long = periodic_on_ranges(Ton_opt, T_off, n_cycles_long)   # same schedule -> same shading for both

model_slow = ABCClonalModel(r_s=r_s_slow, r_r=r_r_slow, carrying_capacity=1000.0, cr_sensitivity=0.0)
y0_slow    = y0_scenario   # canonical names: reused by Part 2 below

t_long, y_long = simulation.simulate_rpa_intermittent(
    model_slow, 'b1', b1_on, b1, t_on=Ton_opt, t_off=T_off,
    n_cycles=n_cycles_long, y0=y0_slow, plot=False)
plot_abc_clonal_run(t_long, y_long, on_ranges_long,
                    title='Scenario A: fully resistant clone (cr_sensitivity=0)')

Cs_long, Cr_long = y_long[2], y_long[3]
total_long = Cs_long + Cr_long
ttp_long = time_to_progression(t_long, total_long, baseline)
print(f'min total = {total_long.min():.4f} at t={t_long[total_long.argmin()]:.1f}')
print(f'final (t={t_long[-1]:.0f}): total={total_long[-1]:.4f}, '
     f'resistant fraction={Cr_long[-1]/total_long[-1]:.4f}, '
     f'time to progression={ttp_long}')
```

### Scenario B — partially resistant clone

$C_r$ is still killed while $B<\mu$, at 50% of $C_s$'s rate (`cr_sensitivity=0.5`). Note this is strong enough that $C_r$'s *own* per-cycle log-growth, $r_r(T_{on}+T_{off}) - \text{cr\_sensitivity}\cdot S(T_{on})$, is actually negative here (≈$0.03\times19.04 - 0.5\times1.83 \approx -0.34$) — so this isn't "resistant cells still growing, just slower"; it's **native feedback strong enough to suppress even the resistant clone**, a stronger and different claim than a small partial-resistance case would give. Same $r_s, r_r$, same initial condition, same schedule, same horizon as Scenario A.

```python
cr_sens_partial = 0.5   # resistant clone still killed, at 50% of Cs's rate
model_partial = ABCClonalModel(r_s=r_s_slow, r_r=r_r_slow, carrying_capacity=1000.0,
                               cr_sensitivity=cr_sens_partial)

t_partial, y_partial = simulation.simulate_rpa_intermittent(
    model_partial, 'b1', b1_on, b1, t_on=Ton_opt, t_off=T_off,
    n_cycles=n_cycles_long, y0=y0_scenario, plot=False)
plot_abc_clonal_run(t_partial, y_partial, on_ranges_long,
                    title=f'Scenario B: partially resistant clone (cr_sensitivity={cr_sens_partial})')

Cs_partial, Cr_partial = y_partial[2], y_partial[3]
total_partial = Cs_partial + Cr_partial
ttp_partial = time_to_progression(t_partial, total_partial, baseline)
print(f'min total = {total_partial.min():.4g} at t={t_partial[total_partial.argmin()]:.1f}')
print(f'final (t={t_partial[-1]:.0f}): total={total_partial[-1]:.4g}, '
     f'resistant fraction={Cr_partial[-1]/total_partial[-1]:.4f}, '
     f'time to progression={ttp_partial}')
```

### Finding the optimal $T_{on}$ for Scenario B — sensitive vs. resistant

Reproduce `figure4 copy.ipynb` Panel C's growth-vs-decay picture, but now for *each* clone separately. $C_s$'s decay integral is the familiar $S(T_{on})$ at full kill strength; $C_r$'s is the same integral scaled by `cr_sensitivity` (its kill is `cr_sensitivity` times as efficient). Growth is $r_s(T_{on}+T_{off})$ for $C_s$ and $r_r(T_{on}+T_{off})$ for $C_r$. Each has its own optimal $T_{on}$ (maximizing net decay) and its own pair of regime-boundary crossings.

```python
def plot_growth_vs_decay(Ton_arr, growth_arr, decay_arr, title='', ax=None):
    """figure4 Panel C style: growth vs. decay curves, 3-regime shading,
    crossing-point markers, and a vertical line at the net-decay-maximizing Ton.
    Returns (Ton_opt, crossings)."""
    net_arr = decay_arr - growth_arr
    sign_changes = np.where(np.diff(np.sign(net_arr)))[0]
    crossings = [brentq(lambda t: np.interp(t, Ton_arr, net_arr), Ton_arr[i], Ton_arr[i + 1])
                for i in sign_changes]
    Ton_opt = Ton_arr[np.argmax(net_arr)]

    standalone = ax is None
    if standalone:
        fig, ax = plt.subplots(figsize=(6.5, 4.2))

    x_max = Ton_arr[-1]
    if len(crossings) >= 2:
        c1x, c2x = crossings[0], crossings[1]
        ax.axvspan(Ton_arr[0], c1x, color='C3', alpha=0.08, lw=0)
        ax.axvspan(c1x, c2x, color='C0', alpha=0.08, lw=0)
        ax.axvspan(c2x, x_max, color='C3', alpha=0.08, lw=0)
    for cx in crossings:
        ax.axvline(cx, color='gray', ls='--', lw=0.8, alpha=0.6)
        ax.plot(cx, np.interp(cx, Ton_arr, growth_arr), 'o', color='k', ms=6, zorder=6, mec='none')

    ax.plot(Ton_arr, growth_arr, color='C3', lw=2, label='growth')
    ax.plot(Ton_arr, decay_arr, color='C0', lw=2, label='decay')
    y_top = max(np.interp(Ton_opt, Ton_arr, growth_arr), np.interp(Ton_opt, Ton_arr, decay_arr))
    ax.plot([Ton_opt, Ton_opt], [0, y_top], color='k', ls=':', lw=1.2, alpha=0.7,
           label=f'optimal $T_{{on}}$={Ton_opt:.2f}')
    ax.axhline(0, color='k', lw=0.5)
    ax.set_xlabel('$T_{on}$')
    ax.set_ylabel('cumulative effect per cycle')
    ax.legend(fontsize=8, frameon=False, loc='upper left')
    sns.despine(ax=ax)
    ax.set_title(title, fontsize=10)

    if standalone:
        plt.tight_layout()
        plt.show()
    return Ton_opt, crossings


growth_s_arr = r_s_slow * (Ton_arr + T_off)
decay_s_arr  = decay_analytic(Ton_arr)
growth_r_arr = r_r_slow * (Ton_arr + T_off)
decay_r_arr  = cr_sens_partial * decay_analytic(Ton_arr)

Ton_opt_s, crossings_s = plot_growth_vs_decay(Ton_arr, growth_s_arr, decay_s_arr,
                                              title='Scenario B: growth vs. decay for $C_s$ (sensitive)')
Ton_opt_r, crossings_r = plot_growth_vs_decay(Ton_arr, growth_r_arr, decay_r_arr,
                                              title='Scenario B: growth vs. decay for $C_r$ (resistant)')

print(f'Ton_opt_s = {Ton_opt_s:.3f}   crossings_s = {[round(c,3) for c in crossings_s]}')
print(f'Ton_opt_r = {Ton_opt_r:.3f}   crossings_r = {[round(c,3) for c in crossings_r]}')
```

Now overlay both on one plot, highlighting each population's growth/decay crossings (regime boundaries) and its own optimal $T_{on}$.

```python
fig, ax = plt.subplots(figsize=(7.5, 4.5))

# shade only the intersection of the two populations' "effective" windows
# (decay > growth for both simultaneously)
effective_lo = max(crossings_s[0], crossings_r[0])
effective_hi = min(crossings_s[1], crossings_r[1])
ax.axvspan(effective_lo, effective_hi, color='C8', alpha=0.15, lw=0)

ax.plot(Ton_arr, growth_s_arr, color='C3', lw=2, ls='-',  label='$C_s$ growth')
ax.plot(Ton_arr, decay_s_arr,  color='C0', lw=2, ls='-',  label='$C_s$ decay')
ax.plot(Ton_arr, growth_r_arr, color='C3', lw=2, ls='--', label='$C_r$ growth')
ax.plot(Ton_arr, decay_r_arr,  color='C0', lw=2, ls='--', label='$C_r$ decay')

for cx in crossings_s:
    ax.plot(cx, np.interp(cx, Ton_arr, growth_s_arr), 'o', color='k', ms=8, zorder=6, mec='none',
           label='_nolegend_')
for cx in crossings_r:
    ax.plot(cx, np.interp(cx, Ton_arr, growth_r_arr), 's', color='k', ms=7, zorder=6, mec='none',
           label='_nolegend_')
# one dummy handle each, just for the legend
# ax.plot([], [], 'o', color='k', ms=8, mec='none', label='$C_s$ crossing')
# ax.plot([], [], 's', color='k', ms=7, mec='none', label='$C_r$ crossing')

# ax.axvline(Ton_opt_s, color='navy',  ls=':', lw=1.4, alpha=0.8, label=f'$C_s$ optimal $T_{{on}}$={Ton_opt_s:.2f}')
# ax.axvline(Ton_opt_r, color='darkorange', ls=':', lw=1.4, alpha=0.8, label=f'$C_r$ optimal $T_{{on}}$={Ton_opt_r:.2f}')
ax.axhline(0, color='k', lw=0.5)

ax.set_xlabel('$T_{on}$')
ax.set_ylabel('cumulative effect per cycle')
ax.set_title('Scenario B: growth vs. decay, sensitive (solid) vs. resistant (dashed)', fontsize=10)
ax.legend(fontsize=7.5, frameon=False, loc='upper left', ncol=2)
sns.despine(ax=ax)
plt.tight_layout()
plt.show()

print(f'Cs effective window: [{crossings_s[0]:.3f}, {crossings_s[1]:.3f}]')
print(f'Cr effective window: [{crossings_r[0]:.3f}, {crossings_r[1]:.3f}]')
print(f'Intersection (effective for both): [{effective_lo:.3f}, {effective_hi:.3f}]')
```

### A second, independent failure mode: growth outracing the maximum achievable decay

Even ignoring $C_r$ entirely, the per-cycle decay $S(T_{on})$ is bounded above (as $T_{on}\to\infty$) by

$$S_\infty = c_2\,\frac{b_2\mu\bigl(1-b_{1,\text{on}}/b_{1,\text{off}}\bigr)}{\lambda_1\lambda_2}$$

If a clone's intrinsic growth rate is comparable to or larger than this bound, growth outpaces the maximum achievable decay for *every* $T_{on}$ — no schedule, however well-timed, can shrink it.

```python
S_inf = -c2_ * K1 * (-1 / lam1 + 1 / lam2)   # decay_analytic(Ton) as Ton -> infinity
print(f'S_inf (max possible log-decay per cycle) = {S_inf:.3f}')

for r_s_test in [0.05, 0.2, 0.5]:
    net_test = decay_analytic(Ton_arr) - r_s_test * (Ton_arr + T_off)
    verdict = 'net decay achievable' if net_test.max() > 0 else 'NO schedule shrinks the tumor'
    print(f'  r_s={r_s_test:4.2f}:  best achievable net decay over all Ton = {net_test.max():+.3f}  ({verdict})')
```

**Takeaway.** The level of resistance — how much the resistant clone still feels the drug — is what actually controls whether the tumor relapses, not the fitness cost (we tried varying $r_r$ alone first; it only changes the *rate* of an outcome that turns out to be inevitable either way — see the log-odds argument above). With $C_r$ completely unaffected by the drug (**Scenario A**, clonal selection dominant), the same schedule that shrinks the tumor to a minimum of ~0.2 by t≈83 can't prevent it regrowing to full carrying capacity by t≈600 (time to progression ≈254) — the untouched resistant clone eventually fills the space vacated by a $C_s$ that keeps getting knocked back every cycle. With `cr_sensitivity=0.5` (**Scenario B**, native feedback dominant), the drug is effective enough against $C_r$ too that both clones are driven toward extinction (total burden ~$10^{-7}$ by t≈600) — resistant *fraction* still creeps toward 100% (fraction alone remains the wrong metric — a residual population this small isn't a clinically meaningful "relapse"), but the tumor itself is essentially eliminated.

A second, independent failure mode doesn't even need a resistant clone: if a clone's own growth rate approaches the bounded maximum per-cycle decay $S_\infty$ (above), no schedule can shrink it at all.

## Part 2 — AB-aware vs. PSA-driven scheduling, across both resistance regimes

The schedule used above (`Ton_opt`, `Toff`) is **open-loop**: it's derived once from the AB kinetics and then applied blindly forever, regardless of how the tumor is actually doing. Real adaptive-therapy trials instead use a **closed-loop, burden-driven** rule — e.g. Zhang et al. (2017)'s prostate-cancer protocol: keep dosing until a measurable biomarker (PSA) drops to some fraction of its pretreatment value, withhold the drug, and resume once PSA regrows back to baseline. This reacts only to the *observable tumor burden* — it has no knowledge of the fast $A,B$ dynamics underneath.

We compare the two philosophies on **both** scenarios from Part 1, since they represent very different regimes:
- **AB-aware**: the fixed periodic schedule from Part 1 (`Ton_opt`, `Toff`), tuned using the AB timescale-separation argument. Already simulated for both scenarios above — no need to rerun.
- **PSA-driven**: treatment is ON until total burden $C_s+C_r$ (our stand-in for PSA) drops to `off_frac` × baseline, then OFF until it regrows to `on_frac` × baseline — repeat. Implemented as `simulation.simulate_threshold_driven`, using `scipy`'s event detection to find the crossing times exactly rather than fixing durations in advance.

### PSA-driven on Scenario A (clonal selection dominant, `cr_sensitivity=0`)

```python
off_frac_ = 0.5
t_psa, y_psa, on_ranges_psa = simulation.simulate_threshold_driven(
    model_slow, 'b1', b1_on, b1, burden_fn, off_frac=off_frac_, on_frac=1.0,
    T_horizon=T_horizon_long, y0=y0_slow)

plot_abc_clonal_run(t_psa, y_psa, on_ranges_psa,
                    title='Scenario A, PSA-driven: treat until burden halves, resume at baseline',
                    thresholds=[('off threshold', off_frac_ * baseline), ('on threshold', 1.0 * baseline)])

Cs_psa, Cr_psa = y_psa[2], y_psa[3]
total_psa = Cs_psa + Cr_psa
print(f'{len(on_ranges_psa)} treatment-on interval(s): {[(round(s,1), round(e,1)) for s,e in on_ranges_psa]}')
print(f'duty cycle (fraction of time on) = {sum(e-s for s,e in on_ranges_psa)/T_horizon_long:.2f}')
print(f'minimum total burden = {total_psa.min():.4g} at t={t_psa[total_psa.argmin()]:.1f}')
print(f'final (t={t_psa[-1]:.0f}): total={total_psa[-1]:.4g}, resistant fraction={Cr_psa[-1]/total_psa[-1]:.4f}')
```

### Is this robust to the threshold choice? (both scenarios)

```python
print(f'{"scenario":10s} {"off_frac":>9s} {"on_frac":>8s} {"n_on":>5s} {"duty":>6s} {"min_total":>12s} '
     f'{"time_to_prog":>13s} {"final_resist_frac":>18s}')
for label, model in [('A', model_slow), ('B', model_partial)]:
    for off_frac, on_frac in [(0.3, 1.0), (0.5, 1.0), (0.7, 1.0), (0.5, 1.5), (0.8, 1.2)]:
        t_i, y_i, on_ranges_i = simulation.simulate_threshold_driven(
            model, 'b1', b1_on, b1, burden_fn, off_frac=off_frac, on_frac=on_frac,
            T_horizon=T_horizon_long, y0=y0_scenario)
        Cs_i, Cr_i = y_i[2], y_i[3]
        total_i = Cs_i + Cr_i
        duty_i  = sum(e - s for s, e in on_ranges_i) / T_horizon_long
        ttp_i   = time_to_progression(t_i, total_i, baseline)
        ttp_str = f'{ttp_i:13.1f}' if ttp_i is not None else f'{"never":>13s}'
        print(f'{label:10s} {off_frac:9.1f} {on_frac:8.1f} {len(on_ranges_i):5d} {duty_i:6.2f} '
             f'{total_i.min():12.4g} {ttp_str} {Cr_i[-1]/total_i[-1]:18.4f}')
```

**Takeaway.** The two scenarios flip which philosophy looks better.

**Scenario A (clonal selection dominant).** PSA-driven settles into a *higher* duty cycle than AB-aware (≈0.68–0.79 across thresholds, vs. AB-aware's fixed ≈0.47) once it starts reacting to a tumor that's regrowing because the drug has gone locally inert against $C_s$ — the same robust-perfect-adaptation wash-out from Part 1. Despite dosing more, it ends up with a far better final resistant fraction than AB-aware's 100% (~5–43%, depending on threshold), though AB-aware still delays progression longest of everything tested (t≈254 vs. t≈203–251) — the same "controls the tumor longest" vs. "preserves the best composition" tension as before.

**Scenario B (native feedback dominant).** The picture reverses. Because the drug is also killing $C_r$ here, burden takes far longer to regrow back to baseline, so PSA-driven settles into a **much lower** duty cycle than AB-aware (≈0.13–0.18 vs. AB-aware's same fixed ≈0.47) — it naturally stops dosing once the tumor is under control, using roughly a third of the drug. And it still gets a *better* final resistant fraction (~2–3%) than AB-aware's ~100% *of a population near numerical extinction* (AB-aware crushes total burden to ~$10^{-7}$, technically "cured" but via massive over-treatment relative to what PSA-driven needed to reach total≈4.8–7.5 and stay there). Neither schedule ever progresses in this regime, so time-to-progression can't distinguish them here — the real difference is how much unnecessary drug the open-loop schedule keeps administering after the job is basically done.

So the open-loop AB-aware schedule is oblivious to which regime it's in: it doses identically (duty≈0.47, forever) whether that's barely enough (Scenario A, where it still loses to full resistance) or wildly excessive (Scenario B, where a closed-loop rule reaches a similar or better outcome on a third of the drug). The PSA-driven controller adapts to the regime automatically, without knowing anything about `cr_sensitivity` — but it isn't uniformly better either, since in Scenario A it needs *more* drug than AB-aware and still can't avoid a substantial resistant fraction. This reinforces the same conclusion as before: the real design target is a controller that combines burden feedback with detecting drug inertness (AB-*and*-burden-aware), since neither open-loop schedule-only nor naive burden-only reasoning gets the dosing right across both regimes.
